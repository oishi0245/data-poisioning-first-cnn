### Resolving `FileNotFoundError` for `archive (1).zip`

To address the `FileNotFoundError`, please ensure your `archive (1).zip` file is uploaded to your Colab session. The most common location for uploaded files is the `/content/` directory. You can upload files using the file browser icon on the left sidebar in Colab.

After uploading, you can verify its presence by running `!ls /content` again. Once the file is confirmed to be in `/content/`, the following code will successfully extract its contents.

```python
import zipfile
import os

# Define the path to your zip file. Assuming it's uploaded to /content/
zip_file_path = '/content/archive (1).zip'

# Define the directory where you want to extract the contents
extraction_dir = 'dataset'

# Create the extraction directory if it doesn't exist
if not os.path.exists(extraction_dir):
    os.makedirs(extraction_dir)

try:
    with zipfile.ZipFile(zip_file_path, 'r') as zip_ref:
        zip_ref.extractall(extraction_dir)
    print(f"Successfully extracted '{zip_file_path}' to '{extraction_dir}'")

    # Optionally, list the contents of the extracted directory to confirm
    print(f"Contents of '{extraction_dir}':")
    !ls {extraction_dir}

except FileNotFoundError:
    print(f"Error: The file '{zip_file_path}' was not found. Please ensure it's uploaded or the path is correct.")
except Exception as e:
    print(f"An error occurred during extraction: {e}")
```

```python
import tensorflow as tf

train_path = "/content/dataset/cifar10/cifar10/train"
test_path = "/content/dataset/cifar10/cifar10/test"

train_dataset = tf.keras.utils.image_dataset_from_directory(
    train_path,
    image_size=(32,32),
    batch_size=32
)

test_dataset = tf.keras.utils.image_dataset_from_directory(
    test_path,
    image_size=(32,32),
    batch_size=32
)
```

```python
import matplotlib.pyplot as plt

class_names = train_dataset.class_names

plt.figure(figsize=(10,10))

for images, labels in train_dataset.take(1):
    for i in range(9):
        ax = plt.subplot(3,3,i+1)
        plt.imshow(images[i].numpy().astype("uint8"))
        plt.title(class_names[labels[i]])
        plt.axis("off")

plt.show()
```
